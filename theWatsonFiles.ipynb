Tokenize dataset

```python
def tokenize(book):
    tokenList = book.split()
    cleanWord = []
    for word in tokenList:
        cleanWord.append(word.strip(string.punctuation).lower())
    return cleanWord
```

Find amount of times Watson is mentioned

```python
def findWatsons(tokenList):
    watsonPos = []
    count = 0
    tokenNum = 0

    for word in tokenList:
        if "watson" in word:
            watsonPos.append(tokenNum)
            count+=1
        tokenNum += 1
    return watsonPos
```
